### Outlier Detection: Sales

I used the Interquartile Range (IQR) method to identify unusually high or low sales values. Outlier detection helps determine whether extreme values should be investigated before further analysis.

````python
Q1 = df["Sales"].quantile(0.25)
Q3 = df["Sales"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

sales_outliers = df[
    (df["Sales"] < lower_bound) |
    (df["Sales"] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Sales Outliers:", len(sales_outliers))
````

**Finding:** The IQR method identified **5,655 Sales outliers**, with values above the upper bound of **581** considered unusually high.

These values were not automatically removed because high sales may represent legitimate high-value transactions. Removing them without further investigation could remove important business information.

### Deeper Finding: What Could Reduce Profitability?

The previous analysis shows that Technology leads in sales and the Consumer segment leads in profit. However, strong sales do not always result in strong profits.

To understand what may affect profitability, the next chart examines the relationships between Sales, Profit, Quantity, Discount, and Shipping Cost.

````python
numeric_columns = [
    "Sales",
    "Profit",
    "Quantity",
    "Discount",
    "Shipping Cost"
]

correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(10, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title(
    "Relationship Between Key Business Variables",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Business Variables")
plt.ylabel("Business Variables")

plt.tight_layout()
plt.show()
````

### Deeper Finding

The heatmap shows that **Discount and Profit have a negative correlation of -0.32**, meaning that higher discounts tend to be associated with lower profits.

It also shows that **Sales and Shipping Cost have the strongest positive relationship at 0.77**, meaning that higher sales tend to be associated with higher shipping costs.

The key business concern is the relationship between discounts and profit. This suggests that the company should review its discount strategy to make sure discounts are supporting sales without reducing profitability too much.

### Deeper Analysis: Average Profit by Discount Level

The correlation analysis showed a negative relationship between Discount and Profit. To understand this relationship more clearly, discounts are grouped into ranges and the average profit is compared across each range.

````python
discount_bins = [0, 0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]

discount_labels = [
    "0-10%",
    "11-20%",
    "21-30%",
    "31-40%",
    "41-50%",
    "51-60%",
    "61-70%",
    "71-80%",
    "81-90%"
]

df["Discount Band"] = pd.cut(
    df["Discount"],
    bins=discount_bins,
    labels=discount_labels,
    include_lowest=True
)

profit_by_discount = (
    df.groupby("Discount Band", observed=False)["Profit"]
    .mean()
    .round(2)
)

print(profit_by_discount)
````

### Visualizing Average Profit by Discount Level

The following bar chart compares average profit across different discount ranges. It helps identify where average profit changes from positive to negative and how profitability varies as discount levels increase.

````python
plt.figure(figsize=(12, 6))

bars = plt.bar(
    profit_by_discount.index.astype(str),
    profit_by_discount.values
)

plt.axhline(0, color="black", linewidth=1)

plt.title(
    "Average Profit by Discount Level",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Discount Level")
plt.ylabel("Average Profit")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

### Deeper Finding: How Discounts Affect Profitability

The analysis shows that average profit generally decreases as discount levels increase.

Transactions with discounts of 0–10% generated an average profit of **62.60**, while those with discounts of 11–20% generated **27.61**.

Average profit became negative in the **21–30% discount band (-21.88)** and remained negative across all higher discount bands.

The 81–90% discount band recorded the largest average loss of **-1,534.33**. However, this band contained only **2 transactions**, so this result should be interpreted cautiously.

**Business Insight:** Discounts above 20% are associated with negative average profit in this dataset. The company should investigate these transactions and review its discount strategy to protect profitability.

````python
discount_counts = (
    df.groupby("Discount Band", observed=False)
    .size()
)

print(discount_counts)
````

## 6. Cause, Effect, and Business Impact

**Cause:** Higher discounts are associated with lower profitability. The correlation between Discount and Profit is **-0.32**, and average profit becomes negative in the **21–30% discount band**.

**Effect:** Transactions with discounts above 20% show negative average profit, suggesting that heavy discounting may be contributing to financial losses.

**Business Impact:** Management should review discounts above 20%, evaluate whether they generate enough additional sales to justify their cost, and consider adjusting discount limits to protect profitability.

````python
discount_bins = [0, 0.10, 0.20, 0.30, 0.40, 0.50, 0.60, 0.70, 0.80, 0.90]

discount_labels = [
    "0-10%",
    "11-20%",
    "21-30%",
    "31-40%",
    "41-50%",
    "51-60%",
    "61-70%",
    "71-80%",
    "81-90%"
]

df["Discount Band"] = pd.cut(
    df["Discount"],
    bins=discount_bins,
    labels=discount_labels,
    include_lowest=True
)

profit_by_discount = (
    df.groupby("Discount Band", observed=False)["Profit"]
    .mean()
    .round(2)
)

print(profit_by_discount)
````
